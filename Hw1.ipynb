5. Створіть два одновимірних масиви розміром 5, заповніть їх випадковими цілими числами в діапазоні від 1 до 10 та виконайте на них поелементні операції додавання, віднімання та множення.


```python
vector1 = np.random.randint(1, 11, size=(5, ), dtype=np.int8)
vector2 = np.random.randint(1, 11, size=(5, ), dtype=np.int8)

print(f"{vector1 = }")
print(f"{vector2 = }")

operation_add = vector1 + vector2
print(f"{operation_add = }")

operation_sup = vector1 - vector2
print(f"{operation_sup = }")

operation_multi = vector1 * vector2
print(f"{operation_multi = }")
```
