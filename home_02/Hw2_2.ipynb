Видаліть усі стовпці з пропусками, крім стовпця "Мова програмування"

```python
df = df.loc[:, (~df.isnull().any() | (df.columns == 'Мова програмування'))]
df.head()
```

Видаліть усі рядки у вихідній таблиці за допомогою методу dropna

```python
df = df.dropna()
df.head()
```

Визначте новий розмір таблиці за допомогою методу shape

```python
df.shape
```

Створіть нову таблицю python_data, в якій будуть тільки рядки зі спеціалістами, які вказали мову програмування Python

```python
python_data = df[df['Мова програмування'] == 'Python']
python_data
```

Визначте розмір таблиці python_data за допомогою методу shape

```python
python_data.shape
```

Використовуючи метод groupby, виконайте групування за стовпчиком "Посада"

```python
group = python_data.groupby("Посада")
```

Створіть новий DataFrame, де для згрупованих даних за стовпчиком "Посада", виконайте агрегацію даних за допомогою методу agg і знайдіть мінімальне та максимальне значення у стовпчику "Зарплата.в.місяць"

```python
grouped_salary = group.agg({"Зарплата на місяць": ["min", "max"]})
grouped_salary
```

Створіть функцію fill_avg_salary, яка повертатиме середнє значення заробітної плати на місяць. Використовуйте її для методу apply та створіть новий стовпчик "avg"

```python
def fill_avg_salary(group_data):
  return pd.Series({
      "avg": group_data['Зарплата на місяць'].sum() / group_data['Зарплата на місяць'].shape[0],
  })

avg = group.apply(fill_avg_salary)

# Flatten the multi-index columns
grouped_salary.columns = [' '.join(col).strip() for col in grouped_salary.columns.values]

grouped_salary = grouped_salary.join(avg)
grouped_salary
```

Створіть описову статистику за допомогою методу describe для нового стовпчика.

```python
grouped_salary['avg'].describe()
```

# Графіки

```python
import seaborn as sns
from matplotlib.ticker import MaxNLocator

palette_name = 'flare'
```

```python
# @title Кому найчастіше підвишують зарплату

plt.figure(figsize=(12, 10))
sns.stripplot(
    y='Мова програмування',
    x='Зміна зарплати за 12 місяців',
    hue='Стать',
    dodge=True,
    palette='flare',
    data=df
  )
plt.show()
```

```python
# @title Розприділення віку по статі

flare_palette = sns.color_palette(palette_name, n_colors=2)

male_ages = df[df['Стать'] == 'чоловіча']['Вік']
female_ages = df[df['Стать'] == 'жіноча']['Вік']

plt.figure(figsize=(12, 6))
plt.hist(male_ages, bins=20, color=flare_palette[0], alpha=0.6, label='Male', edgecolor='black')
plt.hist(female_ages, bins=20, color=flare_palette[1], alpha=0.6, label='Female', edgecolor='black')

plt.title('Age Distribution by Gender')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.legend()

plt.show()
```

```python
# @title Зарплата на місяць vs Зміна зарплати за 12 місяців

flare_palette = sns.color_palette("flare", n_colors=2)

plt.figure(figsize=(12, 6))

sns.stripplot(
    data=df,
    x='Зарплата на місяць',
    y='Зміна зарплати за 12 місяців',
    hue='Стать',
    jitter=True,
    palette=flare_palette,
    dodge=True
)

plt.title('Зарплата на місяць vs Зміна зарплати за 12 місяців')
plt.xlabel('Зарплата на місяць')
plt.ylabel('Зміна зарплати за 12 місяців')

plt.gca().spines[['top', 'right']].set_visible(False)

plt.gca().xaxis.set_major_locator(MaxNLocator(nbins=10))

plt.show()
```
