Створіть нову таблицю python_data, в якій будуть тільки рядки зі спеціалістами, які вказали мову програмування Python

```python
python_data.shape
```

Використовуючи метод groupby, виконайте групування за стовпчиком "Посада"

```python
group = python_data.groupby("Посада")
```

Створіть новий DataFrame, де для згрупованих даних за стовпчиком "Посада", виконайте агрегацію даних за допомогою методу agg і знайдіть мінімальне та максимальне значення у стовпчику "Зарплата.в.місяць"

```python
grouped_salary = group.agg({"Зарплата на місяць": ["min", "max"]})
grouped_salary
```

Створіть функцію fill_avg_salary, яка повертатиме середнє значення заробітної плати на місяць. Використовуйте її для методу apply та створіть новий стовпчик "avg"

```python
def fill_avg_salary(group_data):
  return pd.Series({
      "avg": group_data['Зарплата на місяць'].sum() / group_data['Зарплата на місяць'].shape[0],
  })

avg = group.apply(fill_avg_salary)

# Flatten the multi-index columns
grouped_salary.columns = [' '.join(col).strip() for col in grouped_salary.columns.values]

grouped_salary = grouped_salary.join(avg)
grouped_salary
```
